```python
# FIND-S ALGORITHM WITH HARDCODED DATASET

import pandas as pd

# Hardcoded dataset
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

columns = [
    'Sky',
    'AirTemp',
    'Humidity',
    'Wind',
    'Water',
    'Forecast',
    'EnjoySport'
]

df = pd.DataFrame(data, columns=columns)

print("Dataset:")
print(df)

# Separate features and target
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

# Initialize hypothesis
hypothesis = None

print("\nProcessing Positive Examples:\n")

# Find-S Algorithm
for i in range(len(X)):

    # Consider only positive examples
    if y[i].lower() == 'yes':

        print("Positive Example:", X[i])

        # First positive example initializes hypothesis
        if hypothesis is None:
            hypothesis = X[i].copy()
        else:
            # Compare with current hypothesis
            for j in range(len(hypothesis)):
                if hypothesis[j] != X[i][j]:
                    hypothesis[j] = '?'

        print("Current Hypothesis:", hypothesis)

print("\nFinal Most Specific Hypothesis:")
print(hypothesis)
```

```python
# ============================================
# FIND-S ALGORITHM
# Works with uploaded Kaggle CSV datasets
# ============================================

import pandas as pd
from google.colab import files

# --------------------------------------------
# 1. Upload CSV file
# --------------------------------------------

print("Upload your Kaggle Play Tennis CSV file:")

uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Read CSV
data = pd.read_csv(filename)

print("\nDataset loaded successfully!")
print("Shape:", data.shape)

print("\nColumn Names:")
print(data.columns.tolist())

print("\nFirst 5 rows:")
print(data.head())


# --------------------------------------------
# 2. Preprocessing
# --------------------------------------------

# Remove identifier columns if present
possible_id_columns = ["Day", "day", "ID", "Id", "id"]

for col in possible_id_columns:
    if col in data.columns:
        data = data.drop(col, axis=1)
        print(f"\nRemoved unnecessary column: {col}")


# Remove rows containing missing values
data = data.dropna()

# Remove extra spaces from column names
data.columns = data.columns.str.strip()

# --------------------------------------------
# Automatically select last column as target
# --------------------------------------------

target = data.columns[-1]

# All columns except target are attributes
attributes = list(data.columns[:-1])


print("\nDataset after preprocessing:")
print(data.head())

print("\nNumber of rows:", len(data))

print("\nTarget Column:", target)

print("\nAttributes:")
print(attributes)


# --------------------------------------------
# 3. Initialize Find-S Hypothesis
# --------------------------------------------

# Start with most specific hypothesis
hypothesis = ["∅"] * len(attributes)

print("\nInitial Hypothesis:")
print(hypothesis)


# --------------------------------------------
# 4. Find-S Algorithm
# --------------------------------------------

positive_count = 0

for index, row in data.iterrows():

    # Convert target value to lowercase
    target_value = str(row[target]).strip().lower()

    # Consider only positive examples
    if target_value == "yes":

        positive_count += 1

        print("\n--------------------------------")
        print("Positive Example:", index + 1)
        print("--------------------------------")

        print("Example Values:")

        for attr in attributes:
            print(attr, "=", row[attr])

        # Update hypothesis
        for i in range(len(attributes)):

            # First positive example
            if hypothesis[i] == "∅":
                hypothesis[i] = row[attributes[i]]

            # If values differ, generalize
            elif hypothesis[i] != row[attributes[i]]:
                hypothesis[i] = "?"

        print("\nCurrent Hypothesis:")
        print(hypothesis)


# --------------------------------------------
# 5. Display Final Hypothesis
# --------------------------------------------

print("\n============================================")
print("FINAL HYPOTHESIS")
print("============================================")

if positive_count == 0:
    print("No positive examples ('Yes') found in the dataset.")
    print("\nUnique values in target column:")
    print(data[target].unique())

else:
    print("\nFinal hypothesis attribute-wise:")

    for attribute, value in zip(attributes, hypothesis):
        print(attribute, "=", value)

    print("\nMost Specific Hypothesis:")
    print(hypothesis)

    print("\nNumber of positive examples processed:", positive_count)
```
